```python
import numpy as np
import matplotlib.pyplot as plt
import vector_borne_functions as vbf



T = vbf.T
f = vbf.f
gammas = vbf.gammas
g = vbf.g
chis = vbf.chis
F = vbf.F
daily_mean_arrays = vbf.daily_mean_arrays

dX = vbf.dX
Vector_borne = vbf.Vector_borne

plot_compartments_vs_t = vbf.plot_compartments_vs_t


# GDD^* and n
GDD_R,n = 1500.0, 200


#parameters
Alpha = 1.5
beta = 1.
Gamma = 1.
mu = 1.
delta = mu

parameters = [beta,Alpha,GDD_R,Gamma,mu,delta]


# initial contitions: set the values of I_{H,1}(0) and in I_v(0)
I_H_0 = 0.
I_v_0 = 0.001

initial_conditions= [I_H_0,I_v_0]

#step size: 1/(365*day_fraction)
day_fraction = 5

# T_min and T_max
T_min = 6.66
T_max= 13.10144928


#number of years to simulate
years = 150



#The array X contain the dynamics of each compartment. Rows 0 to n contains the dynamics of the n infected compartments, row n+1 contain te dynamics of the S_H compartment
#row n+2 the dynamics of the R_H compartment, row n+3 the dynamics of the S_v compartment and row n+4 the dynamic of the I_v compartment.
X,times=Vector_borne(parameters,initial_conditions,years,T_min,T_max,n,day_fraction)


fig = plot_compartments_vs_t(X,times,show=True)






#plotting some infected compartments
selected_compartments = [0,1,2]
l_selected = len(selected_compartments)
colors = plt.cm.Spectral_r(np.linspace(0, 1, l_selected))

c = 0
for i in selected_compartments:
    plt.plot(times,X[i],color = colors[c],label = f'$I_{{H,{i}}}$')
    c +=1


plt.xlabel('t',fontsize = 17)
plt.ylabel('$I_{H,i}$', fontsize = 17)
plt.legend()
plt.show()
```
